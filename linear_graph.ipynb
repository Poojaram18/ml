```python
import matplotlib.pyplot as plt

x_values = []
y_values = []

# Generate values for y = x from 0 to 10
for x in range(11):
  x_values.append(x)
  y_values.append(x)

plt.plot(x_values, y_values, marker="o")
plt.title("Linear Function: y = x")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
